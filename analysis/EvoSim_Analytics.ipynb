# EvoSim Analysis Dashboard

This notebook analyzes the `simulation_log.csv` and `world_log.csv` files generated by EvoSim.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Data
sim_file = '../data/demo/simulation_log.csv'
world_file = '../data/demo/world_log.csv'

try:
    sim = pd.read_csv(sim_file)
    world = pd.read_csv(world_file)
    print(f"Loaded {len(sim)} simulation rows")
    print(f"Loaded {len(world)} world rows")
except FileNotFoundError:
    print("Demo data not found. Please run a simulation first.")
```

## 1. Population Dynamics
How does the population stabilize over time?

```python
plt.figure(figsize=(12, 5))
plt.plot(world['Tick'], world['Pop'], label='Population', color='cyan')
plt.fill_between(world['Tick'], world['Pop'], color='cyan', alpha=0.1)
plt.title('Population Over Time')
plt.xlabel('Years')
plt.ylabel('Count')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

## 2. Tribe Formation
Are agents banding together into tribes?

```python
# Calculate percentage of agents in a tribe per tick
tribe_rate = sim.groupby('Tick').apply(lambda x: (x['TribeId'] != -1).mean() * 100)

plt.figure(figsize=(12, 5))
plt.plot(tribe_rate.index, tribe_rate.values, color='orange')
plt.title('Tribe Adoption Rate')
plt.ylabel('% of Agents in Tribes')
plt.xlabel('Years')
plt.grid(True, alpha=0.3)
plt.show()
```

## 3. Karma Distribution
Is the population evolving towards good (positive karma) or evil (negative karma)?

```python
final_tick = sim['Tick'].max()
final_pop = sim[sim['Tick'] == final_tick]

plt.figure(figsize=(10, 6))
sns.histplot(final_pop['Karma'], kde=True, color='purple', bins=20)
plt.title(f'Karma Distribution at Year {final_tick}')
plt.xlabel('Karma')
plt.axvline(0, color='white', linestyle='--', alpha=0.5)
plt.show()
```

## 4. Archetype Survival
Which soul archetypes are most successful?

```python
archetype_counts = final_pop['Archetype'].value_counts()

plt.figure(figsize=(10, 6))
sns.barplot(x=archetype_counts.index, y=archetype_counts.values, palette='viridis')
plt.title(f'Dominant Archetypes at Year {final_tick}')
plt.ylabel('Count')
plt.show()
```
